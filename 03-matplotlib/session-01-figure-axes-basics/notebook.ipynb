## 1. Build It — Core Code

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

### 1.1 A first object-oriented plot

We create a `Figure` and an `Axes` together with `plt.subplots()`,
then draw into the axes with `ax.plot()`.

```python
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

### 1.2 Adding a title, labels, legend, and grid

The methods are all named `ax.set_<property>()`.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), color="tab:blue", linewidth=2.5, label="sin(x)")
ax.plot(x, np.cos(x), color="tab:orange", linestyle="--", label="cos(x)")

ax.set_title("Sine and cosine")
ax.set_xlabel("x (radians)")
ax.set_ylabel("amplitude")
ax.legend(loc="upper right")
ax.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

### 1.3 Format strings, limits, and line styles

```python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), "r--")          # red dashed line
ax.plot(x, np.cos(x), color="green", linewidth=1.5, linestyle="-.")

ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.2, 1.2)
ax.set_title("Explicit limits")
plt.show()
```

### 1.4 Logarithmic axis scales

On a log-log axis a power law appears as a straight line.

```python
t = np.linspace(0.1, 100, 200)

fig, ax = plt.subplots()
ax.plot(t, t ** 2, color="tab:purple", linewidth=2, label="x²")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_title("Power law on log–log axes")
ax.set_xlabel("x (log scale)")
ax.set_ylabel("x² (log scale)")
ax.grid(True, which="both", linestyle=":", alpha=0.5)
ax.legend()
plt.show()
```

### 1.5 Saving a figure

`savefig` writes the figure to disk. The extension picks the format.

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, np.sin(x), color="tab:blue", linewidth=2)
ax.set_title("Saved figure")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
fig.savefig("session01_saved.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 2. Experiment

Predict the result *before* running each cell.

**Experiment 1:** what happens if you change `linewidth` and `linestyle`?

```python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), linewidth=5, linestyle=":", color="crimson")
ax.set_title("linewidth=5, linestyle=':'")
plt.show()
```

**Experiment 2:** narrow the visible window with `set_xlim`.

```python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x))
ax.set_xlim(0, np.pi)          # only one half-period is visible
ax.set_title("xlim = [0, π]")
plt.show()
```

**Experiment 3:** turn the grid on only for the x direction.

```python
fig, ax = plt.subplots()
ax.plot(x, np.sin(x))
ax.grid(True, axis="x", linestyle="-", alpha=0.7)
ax.set_title("Grid on x only")
plt.show()
```

**Experiment 4:** markers make individual samples visible.

```python
sample_x = np.linspace(0, 2 * np.pi, 12)
fig, ax = plt.subplots()
ax.plot(sample_x, np.sin(sample_x), marker="o", linestyle="-", color="tab:green")
ax.set_title("Markers at each sample")
plt.show()
```

**Experiment 5:** compare a linear and a logarithmic y-axis side by side.

```python
fig, ax = plt.subplots()
ax.plot(t, t ** 3)
ax.set_title("Linear y-axis hides the small values")
plt.show()

fig, ax = plt.subplots()
ax.plot(t, t ** 3)
ax.set_yscale("log")
ax.set_title("Log y-axis reveals the small values")
plt.show()
```

## 3. Mini Project — Plot a Function

A polished, saved plot of a damped sine wave.

```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 500)
y = np.sin(x) * np.exp(-x ** 2 / 20)

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(
    x,
    y,
    color="tab:blue",
    linewidth=2.5,
    label="sin(x) · e^(-x²/20)",
)

ax.set_title("Damped sine wave")
ax.set_xlabel("x (radians)")
ax.set_ylabel("amplitude")
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(loc="upper right")

# A little breathing room around the curve
ax.set_xlim(-2 * np.pi, 2 * np.pi)
ax.set_ylim(-0.6, 0.6)

# Reference lines at the axes
ax.axhline(0, color="gray", linewidth=0.8)
ax.axvline(0, color="gray", linewidth=0.8)

fig.savefig("function_plot.png", dpi=150, bbox_inches="tight")
plt.show()
```
