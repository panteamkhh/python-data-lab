```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

## Exercise 1 (Easy) — Plot `y = x²` with labels and grid

```python
x = np.linspace(-5, 5, 400)
y = x ** 2

fig, ax = plt.subplots()
ax.plot(x, y, color="tab:blue", linewidth=2)
ax.set_title("Quadratic function")
ax.set_xlabel("x")
ax.set_ylabel("y = x²")
ax.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

## Exercise 2 (Easy) — `sin` and `cos` on one axes with a legend

```python
x = np.linspace(0, 2 * np.pi, 300)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), color="tab:blue", linestyle="-", label="sin(x)")
ax.plot(x, np.cos(x), color="tab:orange", linestyle="--", label="cos(x)")
ax.set_title("Sine and cosine")
ax.set_xlabel("x (radians)")
ax.set_ylabel("amplitude")
ax.legend(loc="upper right")
plt.show()
```

## Exercise 3 (Medium) — Log y-axis and save to PNG

```python
x = np.linspace(0, 10, 300)
y = 2.0 ** x

fig, ax = plt.subplots()
ax.plot(x, y, color="tab:purple", linewidth=2)
ax.set_yscale("log")
ax.set_title("Exponential growth (log y-axis)")
ax.set_xlabel("x")
ax.set_ylabel("2^x (log scale)")
fig.savefig("exponential.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Exercise 4 (Medium) — Conceptual: pyplot vs. object-oriented API

`plt.plot(x, y)` is the **pyplot state-machine** interface: it draws on the
"current" axes, which Matplotlib tracks implicitly. `fig, ax = plt.subplots()`
followed by `ax.plot(x, y)` is the **object-oriented** interface: it draws on
the exact axes object you named.

Two concrete advantages of the OO style:

1. **Explicitness / determinism** — when several subplots exist, you always know
   which axes you are drawing on; there is no hidden "current axes" to keep track of.
2. **Reusability** — the `fig`/`ax` handles can be passed to functions and
   configured independently, which is essential for building reusable plotting
   helpers and multi-panel figures.

## Exercise 5 (Advanced) — A reusable `plot_function` helper

```python
def plot_function(func, xmin, xmax, title):
    """Create a styled plot of func over [xmin, xmax]; return (fig, ax)."""
    x = np.linspace(xmin, xmax, 400)
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(x, func(x), color="tab:blue", linewidth=2)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("f(x)")
    ax.set_xlim(xmin, xmax)
    ax.grid(True, linestyle="--", alpha=0.5)
    return fig, ax


fig, ax = plot_function(np.sin, -2 * np.pi, 2 * np.pi, "y = sin(x)")
plt.show()

fig, ax = plot_function(np.tanh, -5, 5, "y = tanh(x)")
plt.show()
```

## 🚀 Mini Project — Plot a Function

```python
x = np.linspace(-2 * np.pi, 2 * np.pi, 500)
y = np.sin(x) * np.exp(-x ** 2 / 20)

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(
    x,
    y,
    color="tab:blue",
    linewidth=2.5,
    label="sin(x) · e^(-x²/20)",
)

ax.set_title("Damped sine wave")
ax.set_xlabel("x (radians)")
ax.set_ylabel("amplitude")
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(loc="upper right")

ax.set_xlim(-2 * np.pi, 2 * np.pi)
ax.set_ylim(-0.6, 0.6)

ax.axhline(0, color="gray", linewidth=0.8)
ax.axvline(0, color="gray", linewidth=0.8)

fig.savefig("function_plot.png", dpi=150, bbox_inches="tight")
plt.show()
```
