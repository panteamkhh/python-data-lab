```python
import pandas as pd

df = pd.DataFrame(
    {
        "city": ["Berlin", "Paris", "Berlin", "Lisbon", "Paris", "Berlin"],
        "price": [12.5, 30.0, 7.5, 22.0, 15.0, 40.0],
        "units": [120, 45, 300, 80, 200, 15],
    },
    index=["o1", "o2", "o3", "o4", "o5", "o6"],
)
df
```

## Exercise 1 — `.loc` vs `.iloc` on one value

**(Easy)** `"o3"` is the third label; it sits at position `2`, so `iloc[2, 1]` and `loc["o3", "price"]` agree.

```python
by_label = df.loc["o3", "price"]
by_position = df.iloc[2, 1]
print("loc  :", by_label)
print("iloc :", by_position)
print("equal:", by_label == by_position)
```

## Exercise 2 — Slices

**(Easy)** The `.loc` slice `"o2":"o4"` includes both endpoints (labels), while `.iloc[0:3]`
follows Python slicing and stops before position 3.

```python
print("loc  o2..o4:", df.loc["o2":"o4"].index.tolist())
print("iloc 0..3  :", df.iloc[0:3].index.tolist())
```

## Exercise 3 — Combining conditions

**(Medium, coding)** Wrap each condition in parentheses and join them with `&` / `|`.

```python
and_filter = df[(df["price"] > 15) & (df["units"] < 100)]
print("price > 15 AND units < 100:")
print(and_filter)

or_filter = df[df["city"].isin(["Berlin", "Lisbon"]) | (df["price"] < 10)]
print("\ncity in [Berlin, Lisbon] OR price < 10:")
print(or_filter)
```

## Exercise 4 — `between` and `query`, sorted by units descending

**(Medium)** `between(10, 25)` and `query("10 <= price <= 25")` select the same rows.

```python
between_filter = df[df["price"].between(10, 25)]
print("between(10, 25):")
print(between_filter)

query_filter = df.query("10 <= price <= 25")
print("\nquery('10 <= price <= 25') matches:",
      between_filter.index.equals(query_filter.index))

sorted_result = between_filter.sort_values("units", ascending=False)
sorted_result
```

## Exercise 5 — `.loc` assignment vs chained assignment

**(Advanced, conceptual)** `df.loc[mask, "status"] = "high"` targets the DataFrame itself, so it updates the parent.
`df["status"][mask] = "high"` first returns a temporary Series from `df["status"]` and
writes into *that* object; under Copy-on-Write the parent is left untouched.

```python
demo = df.copy()
demo.loc[demo["price"] > 15, "status"] = "high"
print("loc assignment worked:", "status" in demo.columns)
print(demo[["price", "status"]])
```

## 🚀 Mini Project — Sales Explorer

```python
orders = pd.DataFrame(
    {
        "order_id": ["A100", "A101", "A102", "A103", "A104", "A105", "A106", "A107"],
        "city": ["Berlin", "Paris", "Berlin", "Lisbon", "Paris", "Berlin", "Lisbon", "Paris"],
        "product": ["Widget", "Gadget", "Gizmo", "Widget", "Gadget", "Gizmo", "Gadget", "Widget"],
        "price": [12.5, 30.0, 7.5, 22.0, 15.0, 40.0, 18.0, 26.0],
        "units": [120, 45, 300, 80, 200, 15, 60, 25],
    }
)
orders
```

```python
# Step 2 — whole row for one order by label.
orders.loc[orders["order_id"] == "A100"]
```

```python
# Step 3 — positional slices and column selection.
print("last 3 rows:")
print(orders.iloc[-3:, [1, 3]])
print("\nfirst 4 rows:")
print(orders.iloc[:4, [1, 3]])
```

```python
# Step 4 — combined boolean filter.
big_orders = orders[(orders["price"] > 20) & (orders["units"] > 10)]
big_orders
```

```python
# Step 5 — isin and between.
print("Berlin or Lisbon:")
print(orders[orders["city"].isin(["Berlin", "Lisbon"])][["order_id", "city"]])
print("\nPriced 10 to 30:")
print(orders[orders["price"].between(10, 30)][["order_id", "price"]])
```

```python
# Step 6 — tier via .loc assignment (CoW-safe).
orders.loc[orders["price"] >= 25, "tier"] = "premium"
orders.loc[orders["price"] < 25, "tier"] = "standard"
orders[["order_id", "price", "tier"]]
```

```python
# Step 7 — derive revenue, sort, show top 3.
orders["revenue"] = orders["price"] * orders["units"]
top3 = orders.sort_values("revenue", ascending=False).head(3)
top3[["order_id", "city", "revenue"]]
```

```python
# Step 8 — total revenue per city (full groupby arrives in Session 4).
orders.groupby("city")["revenue"].sum().sort_values(ascending=False)
```
