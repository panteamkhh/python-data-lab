## 2. Experiment

Change a parameter, predict the output, then run the cell.

**Experiment 1** — the index drives alignment, not position.

```python
base = pd.DataFrame({"name": ["Ana", "Bo", "Cara"]}, index=[10, 20, 30])
extra = pd.Series([1.0, 2.0, 3.0], index=[30, 20, 10])
base["rank"] = extra          # aligned by label, so the order flips
base
```

**Experiment 2** — `describe()` ignores text unless you ask for everything.

```python
print("numeric only:")
print(df.describe().columns.tolist())
print("\ninclude='all':")
print(df.describe(include="all").columns.tolist())
```

**Experiment 3** — Copy-on-Write: a slice does not mutate the parent.

```python
original = pd.DataFrame({"x": [1, 2, 3]})
slice_view = original["x"]       # a view under CoW
original.loc[0, "x"] = 99        # safe, goes through .loc
print("original after .loc:", original["x"].tolist())
print("earlier slice is unchanged:", slice_view.tolist())
```

**Experiment 4** — `head` and `tail` accept a row count.

```python
big = pd.DataFrame({"n": range(10)})
print("head(3):", big.head(3)["n"].tolist())
print("tail(3):", big.tail(3)["n"].tolist())
```

**Experiment 5** — a dict of lists with mismatched lengths fails loudly.

```python
try:
    pd.DataFrame({"a": [1, 2, 3], "b": [1, 2]})
except ValueError as err:
    print("ValueError:", err)
```

## 3. Mini Project — Student Gradebook

```python
students = pd.DataFrame(
    {
        "student": ["Ana", "Bo", "Cara", "Dan", "Eve"],
        "math": [91, 78, 84, 65, 88],
        "science": [89, 82, 79, 70, 95],
    }
)
students
```

```python
# Derive the average across the two numeric subjects (row-wise).
students["average"] = students[["math", "science"]].mean(axis=1)
```

### Map a helper function over a Series to produce letter grades.

```python
def letter_grade(avg):
    if avg >= 90:
        return "A"
    if avg >= 80:
        return "B"
    if avg >= 70:
        return "C"
    return "F"


students["letter"] = students["average"].map(letter_grade)
students
```

```python
class_average = students["average"].mean()
passing = int((students["average"] >= 70).sum())
print(f"Class average : {class_average:.2f}")
print(f"Students >= 70: {passing} of {len(students)}")
```
