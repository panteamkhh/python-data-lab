```python
m = np.arange(25).reshape(5, 5)
print("full matrix:\n", m)

print("\n(a) last row:      ", m[-1, :])
print("(b) first column:   ", m[:, 0])
print("(c) center 3x3:\n", m[1:4, 1:4])
```

## Exercise 4 (Medium) — why `np.arange(10).reshape(3, 3)` fails

The original array has **10** elements, but a `(3, 3)` shape needs
`3 × 3 = 9`. Because 10 ≠ 9, NumPy raises a `ValueError`: the number of
elements must be conserved by a reshape.

```python
try:
    np.arange(10).reshape(3, 3)
except ValueError as e:
    print("ValueError:", e)
```

## Exercise 5 (Advanced, conceptual) — read vs. mutate

- **Read only, fastest:** `reshape(-1)` or `ravel()`. Both return a *view*
  and avoid copying data, so they are essentially free.
- **Mutate without touching the original:** `flatten()`. It always returns an
  independent *copy*, so writes to the result cannot leak back into the
  source array.

```python
original = np.arange(6).reshape(2, 3)

view = original.ravel()          # shares memory
copy = original.flatten()        # independent

view[0] = 111
copy[0] = 222

print("original after view write: ", original)   # changed
print("original after copy write: ", original)   # unchanged by the copy
```

## 🚀 Mini Project — Matrix Cropper

```python
def crop(matrix, row_start, row_end, col_start, col_end):
    # Extract a rectangular sub-region using 2D slicing
    return matrix[row_start:row_end, col_start:col_end]

matrix = np.arange(36).reshape(6, 6)
print("Full matrix:\n", matrix)

top_left = crop(matrix, 0, 2, 0, 2)
bottom_right = crop(matrix, 3, 6, 3, 6)

print("\nTop-left 2x2:\n", top_left)
print("Bottom-right 3x3:\n", bottom_right)

# Confirm flatten() copy safety
flat_copy = top_left.flatten()
flat_copy[0] = -1
print("\nFlattened copy modified:", flat_copy)
print("Original matrix unaffected:\n", matrix)
```
