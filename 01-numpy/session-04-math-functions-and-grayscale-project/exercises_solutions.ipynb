## Exercise 2 (Easy) — `np.log(-5)` vs `np.emath.log(-5)`

`np.log(-5)` returns `nan` because no real number logarithm of a negative
exists; `np.emath.log(-5)` returns a complex number by working in the
complex plane.

```python
print("np.log(-5):       ", np.log(-5))
print("np.emath.log(-5): ", np.emath.log(-5))
```

## Exercise 3 (Medium, coding) — inspect an image's channels

This uses a generated placeholder so it runs anywhere; swap in your own file.

```python
import os
from PIL import Image

image_path = "profile.png"
if not os.path.exists(image_path):
    h = w = 120
    img = np.dstack([
        np.broadcast_to(np.linspace(0, 255, w, dtype=np.uint8).reshape(1, w), (h, w)),
        np.broadcast_to(np.linspace(0, 255, h, dtype=np.uint8).reshape(h, 1), (h, w)),
        np.full((h, w), 128, dtype=np.uint8),
    ])
    Image.fromarray(img).save(image_path)

img_data = np.asarray(Image.open(image_path))
print("shape:", img_data.shape)
for i, name in enumerate(["Red", "Green", "Blue"]):
    print(f"{name} channel mean:", img_data[:, :, i].mean())
```

## Exercise 4 (Medium) — equal weights instead of luminance weights

```python
weights_luminance = np.array([0.2126, 0.7152, 0.0722])
weights_equal = np.array([1 / 3, 1 / 3, 1 / 3])

gray_luminance = (img_data @ weights_luminance).astype(np.uint8)
gray_equal = (img_data @ weights_equal).astype(np.uint8)

Image.fromarray(gray_equal).save("profile_gray_equal.png")
print("Luminance mean:", gray_luminance.mean())
print("Equal-weight mean:", gray_equal.mean())
print("Equal weights treat all colors the same; luminance matches human perception.")
```

## Exercise 5 (Advanced, conceptual) — why the matmul is valid

`img_data` has shape `(H, W, 3)` and `weights` has shape `(3,)`.
`matmul` treats the last axis (`3`) as the contraction axis, aligning it with
the sole axis of `weights`. Each pixel's 3-vector is dotted with the weight
vector, and that axis is removed — leaving shape `(H, W)`.

```python
print("img shape:", img_data.shape, "| weights shape:", weights_luminance.shape)
print("result shape:", (img_data @ weights_luminance).shape)
```

## 🚀 Mini Project — Black-and-White Threshold Tool

```python
grayscale = (img_data @ weights_luminance).astype(np.uint8)
threshold = 128
bw = np.where(grayscale > threshold, 255, 0).astype(np.uint8)

Image.fromarray(bw).save("profile_bw.png")
print("Saved profile_bw.png | white pixels:", (bw == 255).mean() * 100, "%")
```

```python
# Try other thresholds
for t in (100, 180):
    bw_t = np.where(grayscale > t, 255, 0).astype(np.uint8)
    print(f"threshold={t:>3} -> white {((bw_t == 255).mean() * 100):5.1f}%")
```
