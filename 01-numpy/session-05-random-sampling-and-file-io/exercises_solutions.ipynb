## Exercise 1 (Easy) — 20 Normal samples (mean 50, std 10)

```python
import numpy as np

np.random.seed(0)
samples = np.random.normal(50, 10, size=20)
print("samples:", samples)
print("sample mean:", samples.mean())
print("sample std: ", samples.std())
```

## Exercise 2 (Easy) — count Uniform samples above 0.5

```python
np.random.seed(1)
u = np.random.uniform(0, 1, size=15)
print("samples:", u)
print("above 0.5:", u[u > 0.5])
print("count:", (u > 0.5).sum())
```

## Exercise 3 (Medium, coding) — save/load a 4×4 matrix via `.npz`

```python
np.random.seed(2)
matrix = np.random.normal(0, 1, size=(4, 4))
np.savez("matrix.npz", matrix=matrix)

reloaded = np.load("matrix.npz")["matrix"]
print("identical?", np.array_equal(matrix, reloaded))
print(reloaded)
```

## Exercise 4 (Medium) — dtype after a CSV round-trip

Values are written as text, and text has no notion of "integer", so
`np.loadtxt` reads everything back as `float64`.

```python
floats = np.array([1.5, 2.25, 3.125, 4.75, 5.0])
np.savetxt("floats.csv", floats, delimiter=",")
reloaded_floats = np.loadtxt("floats.csv", delimiter=",")
print("original dtype:", floats.dtype)
print("reloaded dtype:", reloaded_floats.dtype)
```

## Exercise 5 (Advanced, conceptual) — why the sample converges

Each draw is an independent, identically-distributed sample from the same
distribution. Averaging more such draws cancels individual deviations, so the
average's spread shrinks like `1/sqrt(n)` around the true mean. This is the
**Law of Large Numbers**, and it is why larger samples estimate the true
`mean`/`std` more tightly.

```python
np.random.seed(3)
for n in (10, 100, 10_000, 1_000_000):
    s = np.random.normal(50, 10, size=n)
    print(f"n={n:>9}: mean={s.mean():7.3f}  std={s.std():6.3f}")
```

## 🚀 Mini Project — Simulate, Persist, and Round-Trip a Dataset

```python
import os

np.random.seed(7)
true_mean, true_std = 25.0, 2.5
sensor_readings = np.random.normal(true_mean, true_std, size=1000)

print("True mean/std:", true_mean, true_std)
print("Sample mean/std:", sensor_readings.mean(), sensor_readings.std())

np.savez("sensor_data.npz", sensor_readings=sensor_readings)
reloaded = np.load("sensor_data.npz")["sensor_readings"]
print("\n.npz round-trip identical?", np.array_equal(sensor_readings, reloaded))

subset = sensor_readings[:20]
np.savetxt("sensor_subset.csv", subset, header="reading", delimiter=",")
reloaded_subset = np.loadtxt("sensor_subset.csv", delimiter=",")
print("\nReloaded CSV subset:", reloaded_subset)
print("Reloaded dtype:", reloaded_subset.dtype)
```

```python
# Clean up generated files
for f in ("matrix.npz", "floats.csv", "sensor_data.npz", "sensor_subset.csv"):
    if os.path.exists(f):
        os.remove(f)
print("Cleaned up generated files.")
```
