## 1. Build It — Core Code

```python
import numpy as np

# ---------- 1. Boolean indexing ----------
a = np.arange(12).reshape(3, 4)
print("a:\n", a)

mask = a < 5
print("\nmask (a < 5):\n", mask)
print("a[a < 5]:", a[mask])

# Storing a condition separately for readability
condition = a >= 3
print("a[a >= 3]:", a[condition])

# Combining conditions: parentheses are mandatory
print("a[(a > 2) & (a < 11)]:", a[(a > 2) & (a < 11)])
print("a[(a < 2) | (a % 3 == 0)]:", a[(a < 2) | (a % 3 == 0)])
```

```python
# ---------- 2. Element-wise operations ----------
v1 = np.array([1, 3, 2, 7])
v2 = np.array([4, 1, 1, 3])

print("\nv1 + v2:", v1 + v2, "| np.add:", np.add(v1, v2))
print("v1 * v2:", v1 * v2, "| np.multiply:", np.multiply(v1, v2))
print("v1 ** v2:", v1 ** v2, "| np.power:", np.power(v1, v2))
```

```python
# ---------- 3. Matrix multiplication vs element-wise multiplication ----------
x = np.array([[1, 2], [3, 4]])
y = np.array([[3, 4], [1, 5]])

print("\nElement-wise x * y:\n", x * y)
print("True matrix multiplication x @ y:\n", x @ y)
print("np.matmul(x, y):\n", np.matmul(x, y))
```

```python
# ---------- 4. Broadcasting ----------
scalar_times_vector = np.array([1, 2, 3]) * 2
print("\nvector * scalar (broadcast):", scalar_times_vector)

col = np.array([0, 10, 20, 30]).reshape(4, 1)  # shape (4, 1)
row = np.array([0, 1, 2])                       # shape (3,)
print("Broadcast (4,1) + (3,):\n", col + row)
```

```python
# ---------- 5. Real-world broadcasting: normalization ----------
data = np.array([1, 2, 3, 4, 5])
normalized = (data - data.mean()) / data.std()
print("\nOriginal data:", data)
print("Normalized data:", normalized)
print("New mean (~0):", normalized.mean())
print("New std  (~1):", normalized.std())
```

## 2. Experiment

```python
# Experiment 1: forgetting parentheses in combined conditions raises an error
a = np.arange(10)
try:
    result = a[a > 2 & a < 8]   # missing parentheses -> operator precedence bug
except Exception as e:
    print("Error without parentheses:", type(e).__name__, e)

print("Correct version:", a[(a > 2) & (a < 8)])
```

```python
# Experiment 2: element-wise vs matrix multiplication on the SAME matrices
x = np.array([[2, 0], [0, 2]])
y = np.array([[1, 2], [3, 4]])
print("Element-wise x * y:\n", x * y)
print("Matrix mult   x @ y:\n", x @ y)   # very different result!
```

```python
# Experiment 3: incompatible broadcasting shapes
try:
    a3 = np.array([1, 2, 3])
    b3 = np.array([1, 2])
    print(a3 + b3)
except ValueError as e:
    print("Broadcast error:", e)
```

```python
# Experiment 4: broadcasting a (1,4) row against a (3,1) column
row = np.array([[1, 2, 3, 4]])          # shape (1, 4)
col = np.array([[10], [20], [30]])      # shape (3, 1)
print("Row + Col broadcast result, shape:", (row + col).shape)
print(row + col)
```

```python
# Experiment 5: normalization on a larger, more skewed dataset
big_data = np.array([10, 12, 13, 12, 100])  # one outlier
print("Normalized (with outlier):", (big_data - big_data.mean()) / big_data.std())
```

## 3. Mini Project — Grade Filter and Curve

```python
np.random.seed(42)  # reproducibility
scores = np.random.randint(40, 101, size=20)
print("Scores:", scores)

# At-risk students (boolean indexing)
at_risk = scores[scores < 60]
print("\nAt-risk scores (<60):", at_risk, "| count:", at_risk.size)

# Students scoring between 70 and 85 inclusive (combined condition)
mid_range = scores[(scores >= 70) & (scores <= 85)]
print("Scores between 70-85:", mid_range)

# Curve every score by +5, capped at 100 (broadcasting + np.minimum)
curved = np.minimum(scores + 5, 100)
print("\nCurved scores (capped at 100):", curved)

# Normalize original scores (broadcasting)
normalized_scores = (scores - scores.mean()) / scores.std()
print("\nNormalized scores:", normalized_scores)
```
