## Exercise 1 (Easy) — elements divisible by 4

```python
import numpy as np

arr = np.arange(20)
print(arr[arr % 4 == 0])
```

## Exercise 2 (Easy) — less than 3 OR greater than 15

Use the element-wise OR operator `|` and parenthesize each condition.

```python
print(arr[(arr < 3) | (arr > 15)])
```

## Exercise 3 (Medium) — element-wise vs. matrix product

```python
a = np.arange(9).reshape(3, 3)
b = np.arange(9, 18).reshape(3, 3)

print("a:\n", a)
print("b:\n", b)
print("\nElement-wise (a * b):\n", a * b)
print("\nMatrix product (a @ b):\n", a @ b)
```

## Exercise 4 (Medium) — predict the broadcast shape

`(5, 1)` and `(1, 4)` align to `(5, 4)`: the column stretches across 4
columns and the row stretches down 5 rows.

```python
col = np.arange(5).reshape(5, 1)
row = np.arange(4).reshape(1, 4)
print("col shape:", col.shape, "row shape:", row.shape)
print("sum shape:", (col + row).shape)
print(col + row)
```

## Exercise 5 (Advanced, conceptual) — why one fails, one succeeds

Broadcasting aligns shapes **from the right**:

- `(3,)` and `(2,)` → compare `3` and `2`: not equal and neither is `1`,
  so NumPy raises `ValueError`.
- `(2, 3)` and `(3,)` → align as `(2, 3)` vs `(1, 3)`: the trailing `3`
  matches, and the leading `1` is stretched to `2` → result shape `(2, 3)`.

```python
try:
    np.array([1, 2, 3]) + np.array([1, 2])
except ValueError as e:
    print("(3,) + (2,) ->", e)

result = np.array([[1, 2, 3], [4, 5, 6]]) + np.array([1, 2, 3])
print("(2,3) + (3,) -> shape", result.shape)
print(result)
```

## 🚀 Mini Project — Grade Filter and Curve

```python
np.random.seed(42)
scores = np.random.randint(40, 101, size=20)
print("Scores:", scores)

at_risk = scores[scores < 60]
print("\nAt-risk scores (<60):", at_risk, "| count:", at_risk.size)

mid_range = scores[(scores >= 70) & (scores <= 85)]
print("Scores between 70-85:", mid_range)

curved = np.minimum(scores + 5, 100)
print("\nCurved scores (capped at 100):", curved)

normalized_scores = (scores - scores.mean()) / scores.std()
print("\nNormalized scores:", normalized_scores)
```
